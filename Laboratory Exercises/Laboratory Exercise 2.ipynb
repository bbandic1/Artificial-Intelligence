```python
import pandas as pd

file_path = '/content/Report.csv'

# Load the CSV file into a Pandas DataFrame
df = pd.read_csv(file_path)

# Display the first 5 rows of the DataFrame
print("First 5 rows:")
display(df.head(5))

# Display the last 10 rows of the DataFrame
print("Last 10 rows:")
display(df.tail(10))
```

```python
# Convert 'Ispit1' and 'Ispit1_popravni' columns to numeric values
df['Ispit1'] = pd.to_numeric(df['Ispit1'], errors='coerce')
df['Ispit1_popravni'] = pd.to_numeric(df['Ispit1_popravni'], errors='coerce')

# Create a new column 'Ispit1_final' with the better result between Ispit1 and Ispit1_popravni
df['Ispit1_final'] = df[['Ispit1', 'Ispit1_popravni']].max(axis=1, skipna=True)

# Display the updated DataFrame with 'Ispit1_final' column
print("Updated DataFrame with Ispit1_final:")
display(df[['Indeks', 'Ispit1', 'Ispit1_popravni', 'Ispit1_final']])
```

```python
# Convert 'Ispit2' and 'Ispit2_popravni' columns to numeric values
df['Ispit2'] = pd.to_numeric(df['Ispit2'], errors='coerce')
df['Ispit2_popravni'] = pd.to_numeric(df['Ispit2_popravni'], errors='coerce')

# Create a new column 'Ispit2_final' with the better result between Ispit2 and Ispit2_popravni
df['Ispit2_final'] = df[['Ispit2', 'Ispit2_popravni']].max(axis=1, skipna=True)

# Display the updated DataFrame with 'Ispit2_final' column
print("Updated DataFrame with Ispit2_final:")
display(df[['Indeks', 'Ispit2', 'Ispit2_popravni', 'Ispit2_final']])
```

```python
# Drop the old columns for the exams: 'Ispit1', 'Ispit1_popravni', 'Ispit2', 'Ispit2_popravni'
df.drop(columns=['Ispit1', 'Ispit1_popravni', 'Ispit2', 'Ispit2_popravni'], inplace=True)

# Display the updated DataFrame without the old exam columns
print("Updated DataFrame without the old exam columns:")
display(df)
```

# Exercise 3

```python
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris()

# Features (attributes)
X = iris.data

# Labels (target classes)
y = iris.target

# Feature names
feature_names = iris.feature_names

# Target names (class labels)
target_names = iris.target_names

# Print feature and target names
print("Feature names:", feature_names)
print("Target names:", target_names)

# Print the first 5 rows of the feature matrix
print("\nFirst 5 rows of X:\n", X[:5])
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=1
)

# Print the shapes of the resulting splits
print(X_train.shape)
print(X_test.shape)

print(y_train.shape)
print(y_test.shape)
```

New Shapes: The training set will have 105 samples, and the test set will have 45 samples.

Size of Dataset: The Iris dataset is relatively small, with only 150 samples. While it’s perfect for learning and small-scale testing, it might not be large enough for complex machine learning tasks that require large datasets.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn import metrics

# Initialize the k-NN classifier with k=3
classifier_knn = KNeighborsClassifier(n_neighbors=3)

# Train the classifier using the training data
classifier_knn.fit(X_train, y_train)
```

```python
y_pred = classifier_knn.predict(X_test)

print("Accuracy:", metrics.accuracy_score(y_test, y_pred))
```

The accuracy of 97.78% is excellent. This means the k-NN model is performing very well on the Iris dataset, correctly classifying almost 98% of the test samples. For this basic classification task, this level of accuracy is considered very good.

**Is this accuracy satisfactory?**

This accuracy is highly satisfactory for a simple classification problem like Iris. It indicates that the model is learning effectively from the training data and generalizing well to the test data.

**Is the algorithm applicable in the real world?**

Given that the Iris dataset is small and clean, k-NN works well here. For simple, small-scale classification tasks like flower classification, k-NN is certainly applicable. However, for more complex or larger datasets, we might want to explore other algorithms to handle scalability and noise more effectively.

```python
# Test different values of k
k_values = [1, 3, 5, 7, 9]  # Set of k values to test
accuracies = []

for k in k_values:
    # Initialize the classifier with the current value of k
    classifier_knn = KNeighborsClassifier(n_neighbors=k)

    # Fit the model on the training data
    classifier_knn.fit(X_train, y_train)

    # Make predictions on the test set
    y_pred = classifier_knn.predict(X_test)

    # Calculate accuracy
    accuracy = metrics.accuracy_score(y_test, y_pred)
    accuracies.append((k, accuracy))
    print(f"Accuracy for k={k}: {accuracy}")
```

It looks like the accuracy remains constant for all values of k (1, 3, 5, 7, and 9), all yielding an accuracy of 97.78%.

This suggests that, for this dataset, the choice of k in this range doesn't significantly impact the performance of the model. The k-NN algorithm seems to be stable in this case, and the accuracy remains consistent even when we change the number of neighbors.

Normally, you might expect the accuracy to change with varying k values, especially if k is very small or very large. However, the consistency here indicates that the Iris dataset is relatively simple and clean, so it doesn't require fine-tuning of k for good performance.

**What if k is equal to the number of test samples?**

If k were set to the number of test samples, the model would classify all test samples into one class, resulting in poor performance (around 33% accuracy for a 3-class problem like Iris).

```python
from keras.datasets import mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
# Get the shape of the arrays
print(f"Shape of x_train: {x_train.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of x_test: {x_test.shape}")
print(f"Shape of y_test: {y_test.shape}")

# Calculate memory usage
x_train_memory = x_train.nbytes / (1024 ** 2)  # Convert to MB
y_train_memory = y_train.nbytes / 1024  # Convert to KB
x_test_memory = x_test.nbytes / (1024 ** 2)  # Convert to MB
y_test_memory = y_test.nbytes / 1024  # Convert to KB

# Print memory usage
print(f"Memory usage of x_train: {x_train_memory:.2f} MB")
print(f"Memory usage of y_train: {y_train_memory:.2f} KB")
print(f"Memory usage of x_test: {x_test_memory:.2f} MB")
print(f"Memory usage of y_test: {y_test_memory:.2f} KB")
```

```python
train_images = x_train.reshape((x_train.shape[0], 28 * 28))
train_images = train_images.astype('float32') / 255

test_images = x_test.reshape((x_test.shape[0], 28*28))
test_images = test_images.astype('float32') / 255
```

The reshape function is used to change the shape of the input data without changing the data itself. In this case:

*   x_train.shape[0] refers to the number of training samples (images) in the dataset.
*   28 * 28 is the flattened size of each image, because each MNIST image is originally 28x28 pixels. When reshaped, this will transform each 28x28 image into a one-dimensional array of length 784 (28 multiplied by 28).

This reshaping operation flattens the 28x28 images into a single array of 784 elements for each image. It allows the neural network to process the data as a vector rather than a matrix.



The astype function changes the data type of the array elements. In this case:

*   astype('float32') converts the pixel values to the float32 data type. This is important because most neural network frameworks (like Keras) expect floating-point numbers rather than integers, and it ensures that the numbers can represent decimal values.
*   / 255 scales the pixel values from the original range of 0 to 255 to a range between 0 and 1. This normalization is important for neural networks because it helps the model converge more quickly and ensures that all input values are on the same scale.



```python
# Memory usage after reshaping and converting to float32
train_images_memory = train_images.nbytes / (1024 ** 2)  # Convert to MB
test_images_memory = test_images.nbytes / (1024 ** 2)  # Convert to MB

print(f"Memory usage of train_images: {train_images_memory:.2f} MB")
print(f"Memory usage of test_images: {test_images_memory:.2f} MB")
```

```python
from keras import models
from keras import layers

network = models.Sequential()
network.add(layers.Dense(512, activation='relu', input_shape=(28 * 28,)))
network.add(layers.Dense(10, activation='softmax'))
```

```python
network.compile(optimizer='rmsprop',
                loss='categorical_crossentropy',
                metrics=['accuracy'])
```

```python
from keras.utils import to_categorical

train_labels = to_categorical(y_train)
test_labels = to_categorical(y_test)
```

```python
network.fit(train_images, train_labels, epochs=5, batch_size=128)
```

```python
test_loss, test_acc = network.evaluate(test_images, test_labels)
print('Accuracy for the test dataset:', test_acc)
```

The accuracy on the test set is 97.84%.

The accuracy on the training set is 97.43%.

The difference between the training and test accuracy is very small, which means the model generalizes well and is not overly specific to the training data.
