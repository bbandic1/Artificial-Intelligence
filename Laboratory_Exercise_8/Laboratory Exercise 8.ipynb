```python
import sys
import os
from importlib import reload

try:
    notebook_location_dir = os.path.dirname(os.path.abspath(__file__)) 
except NameError:
    notebook_location_dir = os.getcwd()

task_kb_dirname = "Exercise 2" 
kb_actual_path = os.path.join(notebook_location_dir, task_kb_dirname)

print(f"Putanja do '{task_kb_dirname}' direktorijuma (gdje su driver.py i pravila.krb): {kb_actual_path}")

if not os.path.isdir(kb_actual_path):
    print(f"GREŠKA: Direktorijum '{kb_actual_path}' nije pronađen. Molimo kreirajte ga i stavite 'driver.py' i 'pravila.krb' unutra.")
else:
    print(f"SUCCESS: Direktorijum '{kb_actual_path}' pronađen.")
    if kb_actual_path not in sys.path:
        sys.path.append(kb_actual_path)
        print(f"Dodato u sys.path: {kb_actual_path}")

    try:
        module_name = 'driver'
        
        if module_name in sys.modules:
            print(f"Pokušaj ponovnog učitavanja modula '{module_name}'...")
            task_driver_module = reload(sys.modules[module_name])
        else:
            print(f"Pokušaj prvog importovanja modula '{module_name}'...")
            import driver as task_driver_module 
        
        print(f"Modul '{module_name}.py' uspješno importovan/reloadovan.")
        print("-" * 50)
        
        task_driver_module.run_test_scenario(kb_actual_path, True, True)
        print("-" * 50)
        task_driver_module.run_test_scenario(kb_actual_path, True, False)
        print("-" * 50)
        task_driver_module.run_test_scenario(kb_actual_path, False, False)
        print("-" * 50)
        task_driver_module.run_test_scenario(kb_actual_path, False, True)
        print("-" * 50)

    except ImportError as e:
        print(f"GREŠKA prilikom importovanja '{module_name}.py': {e}", file=sys.stderr)
        print(f"Provjerite da li '{module_name}.py' postoji u '{kb_actual_path}' i da li je ta putanja u sys.path.", file=sys.stderr)
    except Exception as e:
        print(f"Došlo je do neočekivane greške: {e}", file=sys.stderr)
        import traceback
        traceback.print_exc(file=sys.stderr)
```
